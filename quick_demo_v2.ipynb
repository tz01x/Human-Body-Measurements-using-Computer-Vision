```python
import os, subprocess, time, urllib.request, json

REPO    = '/content/Human-Body-Measurements-using-Computer-Vision'
STORAGE = '/content/drive/MyDrive/measure_storage'
PORT    = 8082
TOKEN   = 'pick-a-long-random-string'
os.makedirs(STORAGE, exist_ok=True)

subprocess.run(['pkill', '-f', 'measure_server.py'])
time.sleep(3)

server = subprocess.Popen(
    ['python', '-u', 'measure_server.py',
     '--port', str(PORT), '--storage', STORAGE],
    cwd=REPO, env=dict(os.environ, API_TOKEN=TOKEN),
    stdout=open('/content/server.log', 'w'), stderr=subprocess.STDOUT)
```

```python

for _ in range(90):
    time.sleep(5)
    if server.poll() is not None:
        print(open('/content/server.log').read()[-3000:])
        raise SystemExit('server died')
    try:
        r = json.load(urllib.request.urlopen(
            'http://localhost:%d/health' % PORT, timeout=5))
        if r.get('status') == 'ok':
            print('up on %d:' % PORT, r); break
    except Exception:
        pass
else:
    print(open('/content/server.log').read()[-2000:])
    raise SystemExit('timed out')
```
